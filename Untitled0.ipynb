```python
import pandas as pd

df = pd.read_csv("/content/Q22_airport_checkpoint.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])

print(df.dtypes)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Total rows after cleaning:", len(df))
```

```python
df["hour"] = df["timestamp"].dt.hour
df["day"] = df["timestamp"].dt.day_name()

df.head()
```

```python
checkpoint_wait = df.groupby("checkpoint")["waiting_time_min"].mean()

print(checkpoint_wait)
```

```python
hourly_wait = df.groupby("hour")["waiting_time_min"].mean()

print(hourly_wait)
```

```python
hourly_wait
```

```python
passenger_by_checkpoint = df.groupby("checkpoint")["passenger_count"].sum()

print(passenger_by_checkpoint)
```

```python
passenger_by_checkpoint = df.groupby("checkpoint")["passenger_count"].sum()

print(passenger_by_checkpoint)
```

```python
df["peak_period"] = df["waiting_time_min"] > 15

print(df[["hour", "waiting_time_min", "peak_period"]].head(10))
```

```python
print("Average Waiting Time:", df["waiting_time_min"].mean())
print("Minimum Waiting Time:", df["waiting_time_min"].min())
print("Maximum Waiting Time:", df["waiting_time_min"].max())
print("Total Passengers:", df["passenger_count"].sum())
```

```python
df.to_csv("/content/airport_checkpoint_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
